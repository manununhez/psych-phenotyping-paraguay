**Objetivo:** verificar entorno y dependencias para reproducibilidad técnica.
**Rol en el flujo:** APÉNDICE.
**Entradas requeridas:**
- `requirements.txt`.
**Salidas esperadas:**
- validaciones de entorno en notebook.


```python
import sys
from pathlib import Path

# Añadir la carpeta principal del proyecto al sistema para poder cargar funciones compartidas
sys.path.insert(0, str(Path.cwd().resolve()))

from utils_shared import setup_paths, validate_file_exists

# Inicializar las rutas (carpetas) para que todos los notebooks guarden y lean desde los mismos lugares
paths = setup_paths()

print("\n--- Rutas del Proyecto Inicializadas ---")
for key, value in paths.items():
    print(f"  {key}: {value}")
```

```python
# 1. Verificar Dependencias Básicas
# Comprobamos que las librerías necesarias estén disponibles
print("\nVerificando librerías...")

dependencies = [
    'pandas', 'numpy', 'scipy', 'sklearn', 'transformers', 'torch',
    'matplotlib', 'seaborn', 'nltk', 'spacy', 'tqdm'
]

missing = []
for lib in dependencies:
    try:
        __import__(lib)
        print(f"  {lib}")
    except ImportError:
        print(f"  Error: {lib} (No instalado)")
        missing.append(lib)

if missing:
    print(f"\n[ADVERTENCIA] Faltan librerías: {', '.join(missing)}")
    print("Instálelas usando: pip install -r requirements.txt")
else:
    print("\nTodas las librerías están correctas.")
```

```python
# 2. Validar que los datos originales existan
# Queremos asegurarnos de tener el archivo de historias clínicas (ips_raw.csv)
print("\nBuscando archivo crudo...")

raw_file = paths['DATA_PATH'] / 'ips_raw.csv'

try:
    validate_file_exists(raw_file)
    print(f"  {raw_file.name} encontrado.")
except FileNotFoundError:
    print(f"  [ADVERTENCIA] {raw_file.name} no encontrado en la carpeta 'data'. El pipeline no podrá iniciar.")
```
